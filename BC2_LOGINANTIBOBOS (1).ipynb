```python
import hashlib, time, random, string

# Usuario:admin/contraseña:Admin123!
USERS = {
    "admin": "240be518fabd2724ddb6f04eeb1da5967448d7e831c08c8fa822809f74c720a" # Admin123!
}

def sha256(s):
    return hashlib.sha256(s.encode()).hexdigest()

def validar_password(pwd):
    errores = []
    if len(pwd) < 8:           errores.append("Mínimo 8 caracteres")
    if not any(c.isupper() for c in pwd): errores.append("Al menos 1 mayúscula")
    if not any(c.isdigit() for c in pwd): errores.append("Al menos 1 número")
    if not any(c in "!@#$%^&*()" for c in pwd): errores.append("Al menos 1 especial (!@#$%)")
    return errores

def captcha():
    a, b = random.randint(1,20), random.randint(1,10)
    r = input(f"  CAPTCHA: ¿Cuánto es {a} + {b}? → ").strip()
    return r.isdigit() and int(r) == a + b

def login():
    intentos = 3
    bloqueado_hasta = 0

    print("LOGIN ANTI-BOBOS\n")
    while True:
        if time.time() < bloqueado_hasta:
            resto = int(bloqueado_hasta - time.time())
            print(f" Bloqueado. Espera {resto}s...")
            time.sleep(1); continue

        user = input("Usuario: ").strip()
        if not user: print("Usuario vacío\n"); continue

        print("Verifica que eres humano:")
        if not captcha():
            print("CAPTCHA incorrecto\n"); continue

        pwd = input("Contraseña: ").strip()
        errores = validar_password(pwd)
        if errores:
            print("Contraseña no cumple:")
            for e in errores: print(f"   • {e}")
            print(); continue

        if USERS.get(user) == sha256(pwd):
            print(f"\n¡Bienvenido, {user}!")
            break
        else:
            intentos -= 1
            print(f"Credenciales incorrectas. Intentos restantes: {intentos}")
            if intentos == 0:
                bloqueado_hasta = time.time() + 30
                intentos = 3
                print("Demasiados intentos. Bloqueado 30 segundos.\n")
            print()

login()
```
