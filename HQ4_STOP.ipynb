```python
import random, time

CATEGORIAS = ["Animal", "Fruta", "País", "Color", "Marca", "Nombre", "Objeto"]
LETRAS = "ABCDEFGHIJLMNOPRSTV"

DIFICULTAD = {
    "1": ("Fácil", 60),
    "2": ("Normal", 30),
    "3": ("¿Are you crazy?", 10)
}

PALABRAS_MAQUINA = {
    "Animal":  {"A":"Águila","B":"Ballena","C":"Caballo","D":"Delfín","E":"Elefante","F":"Foca","G":"Gorila","I":"Iguana","J":"Jaguar","L":"León","M":"Mono","N":"Nutria","O":"Oso","P":"Panda","R":"Rinoceronte","S":"Serpiente","T":"Tigre","V":"Vaca"},
    "Fruta":   {"A":"Arándano","B":"Banano","C":"Cereza","D":"Durazno","E":"Estrellafruit","F":"Fresa","G":"Guayaba","I":"Icaco","J":"Jaca","L":"Limón","M":"Mango","N":"Naranja","O":"Olivo","P":"Pera","R":"Rambután","S":"Sandía","T":"Tomate","V":"Uva"},
    "País":    {"A":"Argentina","B":"Bolivia","C":"Colombia","D":"Dinamarca","E":"España","F":"Francia","G":"Grecia","I":"Italia","J":"Japón","L":"Líbano","M":"México","N":"Nicaragua","O":"Omán","P":"Perú","R":"Rusia","S":"Suecia","T":"Turquía","V":"Venezuela"},
    "Color":   {"A":"Azul","B":"Beige","C":"Celeste","D":"Dorado","E":"Esmeralda","F":"Fucsia","G":"Gris","I":"Índigo","J":"Jade","L":"Lila","M":"Morado","N":"Negro","O":"Ocre","P":"Plata","R":"Rojo","S":"Salmón","T":"Turquesa","V":"Verde"},
    "Marca":   {"A":"Adidas","B":"BMW","C":"Canon","D":"Dell","E":"Emirates","F":"Ford","G":"Google","I":"Intel","J":"JBL","L":"LG","M":"Mercedes","N":"Nike","O":"Oracle","P":"Pepsi","R":"Rolex","S":"Samsung","T":"Toyota","V":"Visa"},
    "Nombre":  {"A":"Andrés","B":"Beatriz","C":"Carlos","D":"Diana","E":"Eduardo","F":"Fernanda","G":"Gabriela","I":"Iván","J":"Juliana","L":"Laura","M":"Miguel","N":"Natalia","O":"Oscar","P":"Pablo","R":"Ricardo","S":"Sofía","T":"Tomás","V":"Valentina"},
    "Objeto":  {"A":"Aspiradora","B":"Bolígrafo","C":"Calculadora","D":"Destornillador","E":"Escalera","F":"Florero","G":"Gancho","I":"Impresora","J":"Jarrón","L":"Lámpara","M":"Martillo","N":"Nevera","O":"Olla","P":"Pincel","R":"Reloj","S":"Silla","T":"Tijeras","V":"Ventilador"},
}

def jugar():
    print("=== JUEGO DE STOP ===")
    for k,(nombre,t) in DIFICULTAD.items():
        print(f"  {k}. {nombre} ({t}s)")
    modo = input("Elige dificultad: ").strip()
    nombre_modo, tiempo = DIFICULTAD.get(modo, ("Normal", 30))
    print(f"\nModo: {nombre_modo} | Tiempo: {tiempo}s")

    letra = random.choice(LETRAS)
    print(f"\n¡La letra es: {letra}!\n")
    print("Categorías:", " | ".join(CATEGORIAS))
    print(f"Tienes {tiempo} segundos. ¡Escribe tus respuestas!")

    inicio = time.time()
    respuestas_humano = {}
    for cat in CATEGORIAS:
        if time.time() - inicio >= tiempo:
            print("⏰ ¡Tiempo!")
            break
        r = input(f"{cat}: ").strip()
        respuestas_humano[cat] = r

    # Respuestas máquina
    respuestas_maquina = {cat: PALABRAS_MAQUINA[cat].get(letra, "") for cat in CATEGORIAS}

    # Puntaje
    print(f"\n--- Resultados con letra '{letra}' ---")
    pts_h = pts_m = 0
    for cat in CATEGORIAS:
        h = respuestas_humano.get(cat, "").strip().upper()
        m = respuestas_maquina[cat].upper()
        valida_h = h.startswith(letra) and len(h) > 1
        valida_m = m.startswith(letra) and len(m) > 1
        if valida_h and valida_m and h == m:
            ph, pm = 50, 50
        elif valida_h:
            ph, pm = 100, 0
        elif valida_m:
            ph, pm = 0, 100
        else:
            ph, pm = 0, 0
        pts_h += ph; pts_m += pm
        print(f"{cat:10} | Tú: {respuestas_humano.get(cat,'-'):15} ({ph}pts) | Máquina: {respuestas_maquina[cat]:15} ({pm}pts)")

    print(f"\nTú: {pts_h}pts  |  Máquina: {pts_m}pts")
    if pts_h > pts_m: print("🏆 ¡Ganaste!")
    elif pts_m > pts_h: print("🤖 Ganó la máquina.")
    else: print("🤝 ¡Empate!")

    if input("\n¿Jugar de nuevo? (s/n): ").lower() == 's':
        jugar()

jugar()
```
