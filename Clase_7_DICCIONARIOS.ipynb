```python
notas={"Juan Sebastián": [4.0,3.8,4.2],
       "Maxi": [3.0,4.4,3.3],
       "Luna": [3.5,4.0,5.0],
       "Kamyla": [3.0,5.0,3.5]}
```

```python
def promedio_general(notas):
  suma_total=0
  cantidad_notas=0
  for listas_notas in notas.values():
    suma_total += sum(listas_notas)
    cantidad_notas += len(listas_notas)
  return suma_total/cantidad_notas
```

```python
def promedio_estudiante(notas,nombre):
  if nombre in notas:
    return sum(notas[nombre])/len(notas[nombre])
  else:
    return "No se encuentra"
```

```python
print("promedio general:", promedio_general(notas))
```

```python
print("promedio de Juan Sebastián:", promedio_estudiante(notas, "Juan Sebastián"))
```
