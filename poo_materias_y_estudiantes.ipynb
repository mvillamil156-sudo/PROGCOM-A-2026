```python
class Estudiantes():
  def __init__(self, nombre, apellido, ID, carrera, semestre, telefono, direccion):
    self.nombre=nombre
    self.apellido=apellido
    self.ID=ID
    self.carrera=carrera
    self.semestre=semestre
    self.telefono=telefono
    self.direccion=direccion

def corregir_nombre (self, nuevo_nombre):
  self.nombre=nuevo_nombre

def corregir_apellido(self, nuevo_apellido):
  self.apellido=nuevo_apellido

def corregir_carrera(self, nueva_carrera):
  self.carrera=nueva_carrera

def mostrar_informacion(self):
  print("nombre del estudiante")
  print(f"nombre: {self.nombre}")
  print(f"apellido: {self.apellido}")
  print(f"ID: {self.ID}")
  print(f"carrera: {self.carrera}")
  print(f"semestre: {self.semestre}")
  print(f"telefono: {self.telefono}")
  print(f"direccion: {self.direccion}")
```

```python
class Materia:

    def __init__(self, profesor, lista, max_estudiantes, nombre_materia):
        self.profesor = profesor
        self.lista = lista
        self.max_estudiantes = max_estudiantes
        self.nombre_materia = nombre_materia

    def matricular_estudiante(self, estudiante):
        if len(self.lista) < self.max_estudiantes:
            self.lista.append(estudiante)

    def nota(self, estudiante, nota):
        print(estudiante, "sacó", nota)
m1 = Materia(
    "juanse",
    [["luna", "sanchez", "u00186611", "energias", 3]],
    2,
    "programacion")
m1.matricular_estudiante(
    ["kamyla", "villamil", "u00186613", "energias", 2])
print(m1.lista)
m1.nota("luna", 4.5)
```
