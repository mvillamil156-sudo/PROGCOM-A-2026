##Aprendiendo a usar print
Print() es un método que permite imprimir información en consola

```
if (condicion)==verdadera
  print("Pasan cositas")
else:
  print("Pasan otras cositas")
```
El print() puede imprimir lo siguiente:
1. Cualquier tipo de dato
2. Cualquier método o función (su dirección en memoria)
3. Concatenación en texto
4. Operaciones matemáticas

###Concatenación
Concatenar es unir dos tipos diferentes de datos

````python
x=int(input("Ingrese su edad: "))
y=input("Ingrese su nombre: ")
````

````python
#Ejemplo 1
print(y,"tiene una edad de",x)
````

````python
#Ejemplo 2
print(y+" tiene una edad de "+str(x))
````

````python
#Ejemplo 3
print(f"{y} tiene una edad de {x}")
````
